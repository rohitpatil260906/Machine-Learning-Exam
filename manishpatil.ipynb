# Implement RBF and Polynomial Kernel SVM and compare their performance.

```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
```

```python
# Load dataset
df = pd.read_csv("breast_cancer.csv")

# Drop unnecessary columns
df = df.drop(["Unnamed: 0"], axis=1)
```

### Separate Features and Target

```python
# Separate features and target
x = df.drop("target", axis=1)
y = df["target"]

# Split data
x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.30,
    random_state=42
)
```

```python
# Standardize data
scaler = StandardScaler()

x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

### SVM with Polynomial Kernel

```python
poly_svm = SVC(
    kernel="poly",
    degree=3,
    C=1.0,
    gamma="scale"
)

poly_svm.fit(x_train_scaled, y_train)
y_pred_poly = poly_svm.predict(x_test_scaled)
acc_poly = accuracy_score(y_test, y_pred_poly)

print("SVM with Polynomial Kernel")
print("Accuracy:", acc_poly)
```

### SVM with RBF Kernel

```python
rbf_svm = SVC(
    kernel="rbf",
    C=1.0,
    gamma="scale"
)

rbf_svm.fit(x_train_scaled, y_train)
y_pred_rbf = rbf_svm.predict(x_test_scaled)
acc_rbf = accuracy_score(y_test, y_pred_rbf)

print("SVM with RBF Kernel")
print("Accuracy:", acc_rbf)
```

### Performance Comparison

```python
print("Performance Comparison")
print("Polynomial Kernel Accuracy:", acc_poly)
print("RBF Kernel Accuracy:", acc_rbf)

if acc_poly > acc_rbf:
    print("Polynomial Kernel performs better.")
elif acc_rbf > acc_poly:
    print("RBF Kernel performs better.")
else:
    print("Both kernels have the same accuracy.")
```
